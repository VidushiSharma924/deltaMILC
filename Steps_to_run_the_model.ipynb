# <div align="center">***Steps to run the model using Terminal Commands***</div>

## *1. Open the MILC-DP_1.0 folder and change the directory address to the MILC-DP_1.0 folder where the code is placed*

```python
cd MILC-DP_1.0
```

## *2. Creating the environment*

```python
conda env create -f dmilc_environment.yml
```

## *3. Creating the environment*

```python
conda activate dmilc_environment
```

## *4. Training the model*

```python
nohup python run.py --config config.yml --mode both > DP_train.log 2>&1 &
```

## *5. Testing the model / Inference*

```python
nohup python run_inference.py --config config_inference_new_basins_SUBSET_54.yml --mode eval > Dp_inference.log 2>&1 &
```
